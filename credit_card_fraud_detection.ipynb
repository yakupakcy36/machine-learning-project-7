# Credit card fraud detection

This notebook trains logistic regression on anonymized transactions. Download `creditcard.csv` from the [ULB credit card fraud dataset](https://www.kaggle.com/datasets/mlg-ulb/creditcardfraud) and place it next to the notebook. The Python dependencies are `pandas` and `scikit-learn`.

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score, confusion_matrix, precision_score, recall_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

RANDOM_STATE = 2
```

## Split before undersampling

Use a stratified split on the full dataset to preserve the fraud rate in the holdout. Randomly undersample legitimate transactions only in the training partition. Fit the feature scaler and classifier on that sampled training set.

```python
X = data.drop(columns="Class")
y = data["Class"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE
)
```

```python
fraud_train = y_train[y_train == 1]
legit_train = y_train[y_train == 0].sample(n=len(fraud_train), random_state=RANDOM_STATE)
```

```python
y_train_balanced = pd.concat([legit_train, fraud_train])
X_train_balanced = X_train.loc[y_train_balanced.index]
```

```python
pd.DataFrame({
    "Training before undersampling": y_train.value_counts(),
    "Training after undersampling": y_train_balanced.value_counts(),
    "Held-out test": y_test.value_counts(),
}).sort_index().rename(index={0: "Legitimate", 1: "Fraud"})
```

```python
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
```

```python
model.fit(X_train_balanced, y_train_balanced)
```

## Evaluate on the original class distribution

Average precision summarizes how well the model ranks fraudulent transactions. A random ranking has expected average precision close to the test fraud rate. Precision, recall and the confusion matrix describe the default decision threshold.

This is a random holdout from a two-day dataset. A chronological test would be needed to estimate performance on later transactions.

```python
test_scores = model.decision_function(X_test)
test_predictions = model.predict(X_test)

test_prevalence = y_test.mean()
print(f"Test fraud rate        {test_prevalence:.4%}")
print(f"Average precision      {average_precision_score(y_test, test_scores):.4f}")
print(f"Precision              {precision_score(y_test, test_predictions, zero_division=0):.4f}")
print(f"Recall                 {recall_score(y_test, test_predictions, zero_division=0):.4f}")
```

```python
pd.DataFrame(
    confusion_matrix(y_test, test_predictions, labels=[0, 1]),
    index=["Actual legitimate", "Actual fraud"],
    columns=["Predicted legitimate", "Predicted fraud"],
)
```
